# Gradio

**What it is:** A Python library for building interactive web UI demos for machine learning models, without needing any web development knowledge (no HTML/CSS/JavaScript required).

**How it works:** You wrap your model's prediction function with a few lines of Gradio code, specifying input types (text box, image upload, audio recorder etc) and output types (text, label, image etc). Gradio automatically generates a clean web interface around it.

**Key concepts covered:**
- **Interface class** → the core building block, maps inputs → your function → outputs
- **Blocks** → a more flexible, lower-level way to build custom layouts with multiple components, buttons, and interactions (used when Interface's simple layout isn't enough)
- **Sharing demos** → Gradio can generate a temporary public URL so others can access your demo without you deploying it anywhere
- **HuggingFace Hub integration** → demos can be hosted permanently and for free on HuggingFace Spaces

**Purpose:** Bridges the gap between "I have a trained model" and "someone else can actually try it out" — useful for showcasing work, getting feedback, or building internal tools quickly.

# Argilla

**What it is:** A platform for **human-in-the-loop dataset curation and annotation** — provides both a Python SDK and a web-based GUI where people review, label, and clean datasets used for training or fine-tuning models.

**How it works:**
1. You load an existing dataset (raw or already partially labeled) into an Argilla instance
2. Human annotators view each example through the GUI
3. They actively edit the data:
   - Correcting incorrect labels
   - Flagging or removing corrupted/nonsensical examples
   - Filling in missing labels for unlabeled data
   - Ranking or rating multiple model outputs against each other (common in RLHF-style human preference collection)
4. The reviewed dataset is exported — this is a genuinely modified, higher-quality version of the original, not just a reviewed copy

**Who does the annotation:** Access is controlled, not open to random internet users. Typically done by:
- Dedicated hired annotators (in-house or via annotation services like Scale AI, Labelbox)
- Domain experts reviewing specialized data (medical, legal etc)
- The individual practitioner themselves, if working solo

**Setup:** Argilla runs as a self-hosted instance (via Docker) or through a hosted version on HuggingFace Spaces — you connect to it via Python SDK to push/pull datasets programmatically.

**Purpose:** Directly addresses the "garbage in, garbage out" problem in ML — a model fine-tuned on noisy, mislabeled, or low-quality data performs poorly regardless of architecture or technique. Argilla provides the infrastructure to systematically improve data quality through structured human review before that data is used for fine-tuning.
